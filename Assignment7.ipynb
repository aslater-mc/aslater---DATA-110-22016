# Task 1

Task 1.1

Create a pair plot of all features in the Iris dataset, color-coding by species. Include an overall title for the plot.



```python
sns.pairplot(pt, hue="species", diag_kind="hist", )
plt.suptitle("Iris Dataset Pairplot", y=1.02)
plt.show()
```

Task 1.2

Create a bar plot showing the average sepal length for each iris species.



```python
x = 'species'
y = 'petal_width'
data=pt
```

```python
sns.barplot(x='species', y='sepal_length', data=pt, hue='species', estimator=np.median)
plt.xlabel('Species')
plt.ylabel('Sepal Length')
plt.title('Sepal Length of Each Species of Iris')
plt.show()
```

Task 1.3


Create a box plot showing the distribution of petal widths for each species.



```python
label='species'
ax = sns.boxplot(x='species', y='petal_width', data=pt, hue='species')
plt.xlabel('Species')
plt.ylabel('Petal Width')
plt.title('Petal Width Distribution by Each Species of Iris')
plt.show()
```

Task. 1.4

Create a grouped bar plot showing the average sepal and petal lengths by species.



```python
#Sepal and Iris Petal by Species
plt.bar(pt['species'], pt['sepal_length'], label='Sepal Length', color='darkslateblue')
plt.bar(pt['species'], pt['petal_length'], label='Petal Length', color='mediumseagreen')
plt.xlabel('Species', color='darkslateblue')
plt.ylabel('Length', color='darkslateblue')
plt.title('Sepal and Petal Length of Iris by Species of Iris', color='darkslateblue')
plt.legend(['Sepal Length', 'Petal Length'])
plt.show()
```

Task 1.5


Create a violin plot showing the distribution of sepal length for each species.



```python
sns.violinplot(x='species', y='sepal_length', data=pt, inner=None, zorder=3, hue='species')
plt.xlabel('Species')
plt.ylabel('Petal Width')
plt.title('Sepal Length Distribution by Species of Iris')
plt.grid(axis='y', zorder=0)
plt.show()
```
