```python
#Figure Code
figure=(10,5)
plt.figure(figsize=figure)
deaths = titanic[titanic['survived'] == 0]
survivals = titanic[titanic['survived'] == 1]

#Ploting Code
sns.kdeplot(data=deaths,x='fare',fill=True, color='steelblue', label='Deaths')
sns.kdeplot(data=survivals,x='fare',fill=True, color='orange', label='Survivals')

#Formatting Code
plt.title('KDE Plot Showing the Correlation between Fare and Survival Among Titanic Passengers', color='steelblue')
plt.xlabel('Fare', color='steelblue')
plt.ylabel('Passenger Densisties', color='steelblue')
plt.legend()
plt.xlim(0,550)
plt.grid(color='orange', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#Create KDE plots to compare the age distributions of survivors and non-survivors.
#Question: What trends do you notice in the age distribution of survivors compared to those who did not survive? How might age have impacted survival rates?
```

```python
#Figure Code
figsize=(10,5)
plt.figure(figsize=figsize)

#Ploting Code
sns.kdeplot(data=titanic,x='age',hue='survived', zorder=3)

#Formating Code
plt.grid(color='orange', alpha=0.3, zorder=0)
plt.xlabel('Age of Passengers', color='steelblue')
plt.ylabel('Survival Rate', color='steelblue')
plt.legend(labels=['Survived', 'Dead'])
plt.title('KDE Plot Dictating the Distribution between Age and Survival Among Titanic Passengers', color='steelblue')
plt.xticks([0, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60, 65, 70, 75, 80, 90, 100])
plt.tight_layout()
plt.show()
```

```python
#A. Looking at this KDE plot, we can see that there is a higher rate of survival amongst the youth whereas deaths range higher around the twenty-five to thirties age group. I believe that you had a stronger chance of living had you been younger.
```

```python
#Create a pyramid-style histogram (also known as a population pyramid) for age, showing survival data split by sex. If possible, attempt to find or create one.
#Question: What can you infer from the age and sex distribution in relation to survival rates?
```

```python
#Figure Code
figsize=(10,5)
plt.figure(figsize=figsize)
f_data = titanic.loc[titanic['sex'] == 'female']
m_data = titanic.loc[titanic['sex'] == 'male']

#Ploting Code
#MALES
plt.hist(m_data[m_data['survived'] == 1]['age'], bins=15, alpha=1, label='Male Survived', color='orange', edgecolor='chocolate',  orientation='horizontal', zorder=3,weights=[-1]*len(m_data[m_data['survived'] == 1]))
plt.hist(m_data[m_data['survived'] == 0]['age'], bins=15, alpha=0.3,label='Males Dead', edgecolor='orange', color='orange', orientation='horizontal', zorder=3, weights=[-1]*len(m_data[m_data['survived'] == 0]))

#FEMALES
plt.hist(f_data[f_data['survived']==1]['age'], bins=15, edgecolor='steelblue',color='steelblue', zorder=3,alpha=0.2, label='Females Survived', orientation='horizontal')
plt.hist(f_data[f_data['survived']==0]['age'], bins=15, alpha=1,edgecolor='steelblue', color='lightsteelblue', label='Females Dead', orientation='horizontal',zorder=3)

#Formating Code
plt.legend(loc='upper right')
plt.xticks([-80,-60, -40, -20, 0, 20, 40, 60, 80], [80,60, 40, 20, 0, 20, 40, 60, 80])
plt.yticks([0, 10, 20, 30, 40, 60, 50, 70, 80,90, 100])
plt.grid(color='steelblue', alpha=0.1, axis='y', zorder=0)
plt.xlabel('Amount of Passengers Survived/Dead', color='orange')
plt.ylabel('Ages of Passengers', color='steelblue')
plt.ylim(0,100)
plt.show()
```

```python
#A. From this histogram, we can easily compare the survival rates between the men and women of the titanic. At a glance, more men died than women overall. However, younger males had a better chance of survival than older males, the cut-off being around late 20s, early 30s. I believe this is because, at the time, a societal norm that women and children were to be protected hence the whole "women and children" first.

#NOTE- THIS WAS A BEHEMOTH TO DO! Honestly, this alone took me hours to figure out through trial and error (and lots of error). The .loc that comes with pandas is amazing when it comes to really sourcing out a specific part of a dataset to pull from. I'm super proud of how this one in particular came out. Though I will be transparent, GoogleColab's autofil did come in handy for a lot of this.
```
