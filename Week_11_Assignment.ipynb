```python
#To start, I imported the base packages from the Week 11's reposistory.
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import folium
from folium.plugins import HeatMap
import seaborn as sns
```

```python
df = pd.read_csv('/content/BPD_Arrests.csv')
#The data set we're using today is BPD Arrests, a dataset showcasing the amount of arrests and traits regarding each arrest in the Baltimore area by the Baltimore Police Department.
```

```python
df.head()
```

```python
df.info()
```

```python
#We can see from our df.info the main thing we'll focus on, Gender, is a Float64 dtype.
```

```python
#Foundational Code
cleaned_df=df.dropna(subset=['Latitude', 'Longitude'])
map_center=[39.2904, -76.6122]
m = folium.Map(location=map_center, zoom_start=6)

#Variable Code
f_arrests=cleaned_df[cleaned_df['Gender'] == 'F']
m_arrests=cleaned_df[cleaned_df['Gender'] == 'M']

f_arrest_hdata=f_arrests[['Latitude', 'Longitude']].values
m_arrest_hdata=m_arrests[['Latitude', 'Longitude']].values

#Heatmap Gradient Formating
mgc={0.0: 'red', 0.4: 'yellow', 0.6: 'green', 0.8: 'cyan', 1.0: 'blue'}
fgc={0.0: 'blue', 0.4: 'green', 0.6: 'yellow', 0.8: 'orange', 1.0: 'red'}

#-Note- I reversed the colors for each heatmap so that it would be easier to make comparisons at a glance.

#Heatmap Creation
HeatMap(f_arrest_hdata, name='Female_Arrest_Distribution', gradient=fgc).add_to(m)
HeatMap(m_arrest_hdata, name='Male_Arrest_Distribution', gradient=mgc).add_to(m)

folium.LayerControl().add_to(m)

#Heatmap Export
mapexport="ASlater_Sex-Based_Crime_Heatmap.html"
m.save(mapexport)

mapexport

```
